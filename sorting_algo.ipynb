# This is a review of basic sorting algorithms

```python
def test_sorting(sorting_algo = insertion_sort):
    print(sorting_algo([]) == [])
    print(sorting_algo([1]) == [1])
    print(sorting_algo([2, 1]) == [1, 2])
    print(sorting_algo([2, 1, 3, 5, 4, 7, 8, 6]) == [1, 2, 3, 4, 5, 6, 7, 8])
    print(sorting_algo(list(range(40))[::-1]) == list(range(40)))
```

# Insertion sort

```python
def insertion_sort(L):
    for i in range(1, len(L)):
        key = L[i]
        j = i - 1
        while j >= 0 and L[j] > key:
            L[j+1] = L[j]
            j-=1
        L[j+1] = key
    return L
```

```python
test_sorting()
```

# Merge sort

```python
def fuse_sorted_lists(L1, L2):
    """
    L1 and L2 are two sorted lists
    """
    final = []
    curr1, curr2 = 0, 0
    while curr1<len(L1) and curr2<len(L2):
        if L1[curr1] <= L2[curr2]:
            final.append(L1[curr1])
            curr1 += 1
        else:
            final.append(L2[curr2])
            curr2 += 1
    
    final.extend(L1[curr1:])
    final.extend(L2[curr2:])
    return final

def merge_sort(L):
    if len(L)<=1:
        return L

    n = len(L)
    return fuse_sorted_lists(merge_sort(L[n//2:]), merge_sort(L[:n//2]))
```

```python
test_sorting(merge_sort)
```

# Quick sort

```python
def quick_sort(L):
    if len(L)<=1:
        return L
    
    pivot = L[-1]
    left = [elem for elem in L[:-1] if elem<=pivot]
    right = [elem for elem in L[:-1] if elem >pivot]

    return quick_sort(left) + [pivot] + quick_sort(right)
```

```python
test_sorting(quick_sort)
```

# Bucket sort

When data is evenly distributed in an interval

```python
import math

def bucket_sort(L, k):
    """
    L the list and k the number of buckets
    """
    if len(L)==0:
        return L
    buckets = [[] for _ in range(k)]

    x_max = max(L)
    for x in L:
        # Pour x entre 0 et x_max exclu, la condition pourrait être
        buckets[math.floor(x*k/(1.01*x_max))].append(x)

    final = []
    for bucket in buckets:
        final.extend(insertion_sort(bucket))
    
    return final
```

```python
test_sorting(lambda L : bucket_sort(L, 3))
```
